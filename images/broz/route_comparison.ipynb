# Porovnání trasy: Mapy.cz vs. OSMnx

**Trasa:** Nučice → Nové Butovice (Praha 13)

Mapy.cz hledá nejrychlejší cestu (preferuje D5/D0). OSMnx přes Bellman-Forda hledá cestu nejkratší podle délky hran.

```python
import matplotlib.pyplot as plt
import networkx as nx
import osmnx as ox
```

```python
START = (50.0185, 14.2125)  # Nučice
END   = (50.0535, 14.3460)  # Nové Butovice

north = max(START[0], END[0]) + 0.025
south = min(START[0], END[0]) - 0.025
east  = max(START[1], END[1]) + 0.025
west  = min(START[1], END[1]) - 0.025

G = ox.graph_from_bbox(bbox=(west, south, east, north), network_type='drive')
len(G.nodes), len(G.edges)
```

```python
orig = ox.distance.nearest_nodes(G, X=START[1], Y=START[0])
dest = ox.distance.nearest_nodes(G, X=END[1],   Y=END[0])

route   = nx.bellman_ford_path(G, orig, dest, weight='length')
length  = nx.bellman_ford_path_length(G, orig, dest, weight='length')
print(f'Délka: {length/1000:.2f} km, uzlů: {len(route)}')
```

```python
fig, ax = ox.plot_graph_route(
    G, route,
    route_color='#ff2d2d', route_linewidth=3,
    node_size=0, edge_color='#888', edge_linewidth=0.4,
    bgcolor='#000', show=False, close=False,
)
ax.set_title(f'OSM Bellman — délka cesty {int(round(length))} m', color='#aaa')
plt.show()
```

## Porovnání

| Zdroj | Délka | Čas |
|---|---|---|
| Mapy.cz | 15,8 km | 16 min |
| OSMnx (Bellman-Ford) | ~11,9 km | ~14 min @ 50 km/h |

Mapy.cz vede řidiče přes dálnici D5/D0 (delší vzdálenost, vyšší rychlost = kratší čas). OSMnx volí absolutně nejkratší cestu nezávisle na typu silnice, proto kličkuje přes menší silnice.
