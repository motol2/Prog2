```
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Vklad:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Výběr:", amount)
        else:
            print("Nedostatek peněz!")

    def show_balance(self):
        print("Zůstatek na účtu:", self.balance)

# vytvoření objektu
account = BankAccount("Tom", 1000)

account.deposit(500)
account.withdraw(300)
account.show_balance()
```

```
class Player:
    def __init__(self, name, goals):
        self.name = name
        self.goals = goals

    def score_goal(self):
        self.goals += 1
        print(self.name, "dal gól!")

    def show_stats(self):
        print(self.name, "má", self.goals, "gólů.")


# podtřída (dědí z Player)
class Goalkeeper(Player):
    def __init__(self, name, goals, saves):
        super().__init__(name, goals)  # zavolá konstruktor rodiče
        self.saves = saves

    def make_save(self):
        self.saves += 1
        print(self.name, "chytil střelu!")

    def show_stats(self):
        print(self.name, "má", self.goals, "gólů a", self.saves, "zákroků.")


# použití
gk = Goalkeeper("Neuer", 0, 5)

gk.make_save()
gk.make_save()
gk.show_stats()
```

```
class Player:
    def __init__(self, name, number):
        self.name = name
        self.number = number

    def introduce(self):
        print("Jsem hráč", self.name, "s číslem", self.number)


# podtřída útočník
class Striker(Player):
    def __init__(self, name, number, goals):
        super().__init__(name, number)
        self.goals = goals

    def score_goal(self):
        self.goals += 1
        print(self.name, "dal gól!")

    def show_stats(self):
        print(self.name, "má", self.goals, "gólů.")


# podtřída obránce
class Defender(Player):
    def __init__(self, name, number, tackles):
        super().__init__(name, number)
        self.tackles = tackles

    def make_tackle(self):
        self.tackles += 1
        print(self.name, "zastavil soupeře!")

    def show_stats(self):
        print(self.name, "má", self.tackles, "zákroků.")


# podtřída brankář
class Goalkeeper(Player):
    def __init__(self, name, number, saves):
        super().__init__(name, number)
        self.saves = saves

    def make_save(self):
        self.saves += 1
        print(self.name, "chytil střelu!")

    def show_stats(self):
        print(self.name, "má", self.saves, "zákroků.")


# vytvoření objektů
s = Striker("Ronaldo", 7, 10)
d = Defender("Van Dijk", 4, 20)
g = Goalkeeper("Neuer", 1, 5)

# použití
s.score_goal()
d.make_tackle()
g.make_save()

s.show_stats()
d.show_stats()
g.show_stats()
```
